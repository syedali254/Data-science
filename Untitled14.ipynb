```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

import numpy as np
import pandas as pd

np.random.seed(42)

n=200
age1=np.random.normal(24,4,n)
age2=np.random.normal(55,5,n)


age=np.concatenate([age1,age2])

inc1=np.random.normal(40000,5000,n)
inc2=np.random.normal(120000,12000,n)
income=np.concatenate([inc1,inc2])
print(income)
```

```python
df=pd.DataFrame({
    'age':age,
    'income':income
})
print(df.head(5))
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(df)
from sklearn.metrics import silhouette_score
inertia_list=[]
silh_score=[]

k_range=range(2,11)
for k in range(2,11):
    km=KMeans(n_clusters=k,n_init=10,random_state=42)
    km.fit(X_scaled)
    labels=km.predict(X_scaled)
    inertia_list.append(km.inertia_)
    silh_score.append(silhouette_score(X_scaled,labels))


print(silh_score)

plt.plot(k_range,inertia_list)
plt.show()


plt.plot(k_range,silh_score)
plt.show()
```

```python
from sklearn.cluster import KMeans
kme=KMeans(n_clusters=2,n_init=10,random_state=42)
df['cluster_id']=kme.fit_predict(X_scaled)
cluster_profiles = df.groupby('cluster_id').mean()
print(cluster_profiles.round(2))
```

```python
from sklearn.metrics import silhouette_samples
kme1=KMeans(n_clusters=2,n_init=10,random_state=42)
cluster_labels=kme1.fit_predict(X_scaled)
avg_s=silhouette_score(X_scaled,cluster_labels)
print(avg_s)

sample_s=silhouette_samples(X_scaled, cluster_labels)
print(sample_s)
```

```python
from sklearn.metrics import silhouette_samples
colors = ['#14b8a6', '#8b5cf6', '#f43f5e'] # Clean modern palette matches lab hex guides
k_clusters=2
y_lower=10
fig, ax = plt.subplots(figsize=(8, 6))
for i in range(k_clusters):
    # Gather and sort silhouette values for the current cluster
    ith_cluster_values = sample_s[cluster_labels == i]
    ith_cluster_values.sort()
    
    size_cluster_i = ith_cluster_values.shape[0]
    y_upper = y_lower + size_cluster_i
    
    # Fill the silhouette profile shape smoothly
    ax.fill_betweenx(np.arange(y_lower, y_upper), 0, ith_cluster_values,
                     facecolor=colors[i], edgecolor=colors[i], alpha=0.7)
    
    # Label each cluster silhouette shape at its center point
    ax.text(-0.05, y_lower + 0.5 * size_cluster_i, f"Cluster {i}", 
            ha='right', va='center', fontsize=10, fontweight='bold', color='white')
    
    # Shift the bottom starting position for the next cluster shape
    y_lower = y_upper + 10 

# Draw a red dashed reference line showing the global average silhouette score
ax.axvline(x=avg_s, color="red", linestyle="--", linewidth=2, 
           label=f"Global Average: {avg_s:.3f}")

ax.set_title("Silhouette Coefficient Profile Distribution Across Clusters", fontsize=12, fontweight='bold')
ax.set_xlabel("Silhouette Coefficient Value")
ax.set_ylabel("Sorted Data Points Vector Space")
ax.set_yticks([])  # Clear out numeric y-ticks since labels are handled via text
ax.set_xlim([-0.1, 1.0]) # Set window limits to catch misallocated outliers dipping below 0
ax.legend(loc="best")
plt.tight_layout()
plt.show()
```
