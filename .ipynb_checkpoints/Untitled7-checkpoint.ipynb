```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

np.random.seed(42)

num_students=n=200
cgpa=np.random.uniform(1.0,4.0,n)
DSA_skill=np.random.randint(40,101,n)
communication_skill = np.random.randint(35, 101, num_students)

internships = np.random.randint(0, 4, num_students)

projects = np.random.randint(1, 8, num_students)

aptitude_score = np.random.randint(40, 101, num_students)

attendance = np.random.randint(50, 101, num_students)


placement_score = (
    cgpa * 15 +
    DSA_skill * 0.25 +
    communication_skill * 0.20 +
    internships * 10 +
    projects * 5 +
    aptitude_score * 0.20 +
    attendance * 0.05
)

noise=np.random.normal(0,10,n)
score=placement_score+noise
status=np.where(score>85)


df=pd.DataFrame({
    'cgpa':cgpa,
    'DSA_skill':DSA_skill,
    'communication_skill':communication_skill,
    'internships':internships,
    'projects':projects,
    'aptitude_score':aptitude_score,
    'attendance':attendance,
    'status':status,
})



X=df[['cgpa','DSA_skill','communication_skill','projects','aptitude_score','attendance']]
Y=df['status']
    
    
    
    


```
