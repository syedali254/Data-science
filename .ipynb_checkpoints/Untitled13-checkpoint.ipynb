```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier


np.random.seed(42)
n=200
cgpa=np.random.normal(2.9,0.7,n)

Dsa_skill=np.random.uniform(4,10,n)
internships=np.random.randint(1,5,n)
projects=np.random.randint(1,6,n)


placement_score=(cgpa*0.4)+(Dsa_skill*2)+(internships*0.3)+(projects*0.1)
score=placement_score+(np.random.normal(0,1.5,n))
print(score)
```
