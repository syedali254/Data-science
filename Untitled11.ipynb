```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# 1. Synthesize raw data with unscaled, highly mismatched features
np.random.seed(42)
income = np.concatenate([np.random.normal(30000, 5000, 100), np.random.normal(90000, 15000, 100)])
age = np.concatenate([np.random.normal(23, 3, 100), np.random.normal(52, 8, 100)])

df = pd.DataFrame({'Income': income, 'Age': age})

# 2. Resolve feature scaling imbalances using StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)

# 3. Hyperparameter optimization loops
inertia_list = []
silhouette_list = []
k_range = range(2, 11) # Checking cluster sizes from 2 up to 10

for k in k_range:
    km = KMeans(n_clusters=k, n_init=10, random_state=42)
    labels = km.fit_predict(X_scaled)
    
    inertia_list.append(km.inertia_)
    silhouette_list.append(silhouette_score(X_scaled, labels))

# 4. Generate side-by-side metric verification curves
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot the Elbow Curve
axes[0].plot(k_range, inertia_list, marker='o', color='purple')
axes[0].set_xlabel('Number of Clusters (K)')
axes[0].set_ylabel('Inertia (WCSS)')
axes[0].set_title('The Elbow Method Optimization Curve')

# Plot Silhouette Scores
axes[1].plot(k_range, silhouette_list, marker='s', color='teal')
axes[1].set_xlabel('Number of Clusters (K)')
axes[1].set_ylabel('Silhouette Quality Score')
axes[1].set_title('Silhouette Analysis Profile')

plt.tight_layout()
plt.show()
```
