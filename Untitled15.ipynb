```python
import numpy as np

experience = np.array([1, 2, 3, 4, 5])       # x values
salary     = np.array([40, 55, 60, 75, 85]) 


def linear_reg(x,y):
    x_mean=np.mean(x)
    y_mean=np.mean(y)
    numerator=np.sum((x-x_mean)*(y-y_mean))
    den=np.sum((x-x_mean)**2)
    
    
    B1=numerator/den
    print(1)

    B0=y_mean-(B1*x_mean)

    
    return B0,B1



b0,b1=linear_reg(experience,salary)
print(b0,b1)



def predict(x):
    return b0+b1*x


price=predict(10)
print("price is : ",price)

y=salary
y_pred=predict(experience)
error=np.sum((y-y_pred)**2)
print(y)
print(y_pred)
print(error)
print(np.sqrt(error))


new=predict(10)
print(new)
```
