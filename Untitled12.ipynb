```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

np.random.seed(42)
inc1=np.random.normal(30000,40000,100)
inc2=np.random.normal(100000,15000,100)

income=np.concatenate([inc1,inc2])

age1=np.random.normal(24,3,100)
age2=np.random.normal(54,7,100)
age=np.concatenate([age1,age2])

df=pd.DataFrame({
    'income':income,
    'age':age,
})

inertia_list=[]
silh_list=[]

scaler=StandardScaler()
X_scaled=scaler.fit_transform(df)
k_range=range(2,11)
for k in range(2,11):
    km=KMeans(n_clusters=k,n_init=10,random_state=42)
    km.fit(X_scaled)
    labels=km.predict(X_scaled)
    inertia_list.append(km.inertia_)
    silh_list.append(silhouette_score(X_scaled,labels))
    

print(inertia_list)
print(silh_list)




plt.plot(k_range,inertia_list)
plt.show()


plt.plot(k_range,silh_list)
plt.show()
#plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=labels, cmap='viridis', alpha=0.6, label='Data Points')
#plt.show()
```
