```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler

np.random.seed(42)

num_students=n=200
cgpa=np.random.uniform(1.0,4.0,n)
DSA_skill=np.random.randint(40,101,n)
communication_skill = np.random.randint(35, 101, num_students)

internships = np.random.randint(0, 4, num_students)

projects = np.random.randint(1, 8, num_students)

aptitude_score = np.random.randint(40, 101, num_students)

attendance = np.random.randint(50, 101, num_students)



placement_score = (
    cgpa * 15 +
    DSA_skill * 0.25 +
    communication_skill * 0.20 +
    internships * 10 +
    projects * 5 +
    aptitude_score * 0.20 +
    attendance * 0.05
)

# Add random noise
noise = np.random.normal(0, 10, n)

score = placement_score + noise

# Classification target
placed = np.where(score > 85,"Placed","Not")

df = pd.DataFrame({
    'cgpa': cgpa,
    'DSA_skill': DSA_skill,
    'communication_skill': communication_skill,
    'internships': internships,
    'projects': projects,
    'aptitude_score': aptitude_score,
    'attendance': attendance,
    'placed': placed
})
X=df.drop(['placed'],axis=1)
Y=df['placed']

X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)

scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)

knn=KNeighborsClassifier(n_neighbors=4)
knn.fit(X_train_scaled,Y_train)
pred=knn.predict(X_test_scaled)
acc=accuracy_score(Y_test,pred)
print("accuracy is: ",acc)


test_acc=knn.score(X_test_scaled,Y_test)
print("test score is: ",knn.score(X_test_scaled,Y_test))
train_acc=knn.score(X_train_scaled,Y_train)
print("test score is: ",train_acc)






new_student = pd.DataFrame({
    'cgpa': [3.7],
    'DSA_skill': [88],
    'communication_skill': [80],
    'internships': [2],
    'projects': [5],
    'aptitude_score': [85],
    'attendance': [90]
})

new_scaled=scaler.transform(new_student)


predict=knn.predict(new_scaled)


pr=knn.predict_proba(new_scaled)
if predict==1:
    print("placed")
else:
    print("not placed")

print("probability is: ",pr)
    







    
    
    
    


```
