```python

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_csv("loan_data_set.csv")
df.head(3)
```

```python
df.isnull().sum()
```

```python
df["Gender"].fillna(df["Gender"].mode()[0],inplace=True)
```

```python
for i in df.select_dtypes(include="object").columns:
    df[i].fillna(df[i].mode()[0],inplace=True)
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
for i in df.select_dtypes(include="float64").columns:
    df[i].fillna(df[i].mean(),inplace=True)
```

```python
df.isnull().sum()
```

```python
#2Ordinal Endoign when order matters like sizes 

df2=pd.DataFrame({"size":["s","m","s","s","xl","l","xl","m","l"]})
```

```python
df2
```

```python
orders=[['s','m','l','xl']]
```

```python
from sklearn.preprocessing import OrdinalEncoder
oe=OrdinalEncoder(categories=orders)
oe.fit(df2[["size"]])
```

```python
df2["orderd"]=oe.transform(df2[["size"]])
```

```python
df2
```

```python
#ordinal Encoding using Map function

order_d={"s":0,"m":1,"l":2,"xl":3}
df2["order_map"]=df2["size"].map(order_d)
df2
```
