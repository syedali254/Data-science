```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler

np.random.seed(42)

num_students=n=200
cgpa=np.random.uniform(1.0,4.0,n)
DSA_skill=np.random.randint(40,101,n)
communication_skill = np.random.randint(35, 101, num_students)

internships = np.random.randint(0, 4, num_students)

projects = np.random.randint(1, 8, num_students)

aptitude_score = np.random.randint(40, 101, num_students)

attendance = np.random.randint(50, 101, num_students)



placement_score = (
    cgpa * 15 +
    DSA_skill * 0.25 +
    communication_skill * 0.20 +
    internships * 10 +
    projects * 5 +
    aptitude_score * 0.20 +
    attendance * 0.05
)

# Add random noise
noise = np.random.normal(0, 10, n)

score = placement_score + noise

# Classification target
placed = np.where(score > 85,"Placed","Not")

df = pd.DataFrame({
    'cgpa': cgpa,
    'DSA_skill': DSA_skill,
    'communication_skill': communication_skill,
    'internships': internships,
    'projects': projects,
    'aptitude_score': aptitude_score,
    'attendance': attendance,
    'placed': placed
})
X=df.drop(['placed'],axis=1)
Y=df['placed']

X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)

scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)

test_acc=[]
train_acc=[]
k_list=[]
for k in range(1,21,2):
    knn=KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train_scaled,Y_train)
    train_ac=knn.score(X_train_scaled,Y_train)
    train_acc.append(train_ac)
    test_ac=knn.score(X_test_scaled,Y_test)
    test_acc.append(test_ac)
    k_list.append(k)




print(test_acc)
print(train_acc)

import matplotlib.pyplot as plt


print("----------------relation with train data-----------------")

plt.plot(k_list,train_acc)
plt.show()


print("----------------relation with test data-----------------")

plt.plot(k_list,test_acc)
plt.show()
```
