```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score
import pandas as pd
from sklearn.model_selection import train_test_split
np.random.seed(42)
n=500
```

```python
sqft=np.random.randint(800,3500,n)
bedrooms=np.random.randint(2,7,n)
age=np.random.randint(1,30,n)

noise=np.random.normal(0,15,n)
```

```python
price=50+0.08*sqft + 15*bedrooms -1.5*age +noise
```

```python
df=pd.DataFrame({
    'sqft':sqft,
    'bedrooms':bedrooms,
    'age':age,
    'price':price,
})

X=df.drop(['price'],axis=1)
Y=df['price']

X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)

print("Train data: ",len(X_train))
print("test data: ",len(X_test))
model=LinearRegression()
model.fit(X_train,Y_train)

b0=model.intercept_
b1=model.coef_[0]
print("Slope : ",b1)
print("intercept: ",b0)
y_pred=model.predict(X_test)
print("Train data accuracy is: ",model.score(X_train,Y_train))
print("Test data accuracy is: ",model.score(X_test,y_pred))
print("R2 score is: ",r2_score(Y_test,y_pred))
print("mse is: ",mean_squared_error(Y_test,y_pred))
print("rmse is: ",np.sqrt(mean_squared_error(Y_test,y_pred)))

new_house = np.array([[2000, 3, 10]]) 

print("Price of this house is: ",model.predict(new_house)[0])
```
